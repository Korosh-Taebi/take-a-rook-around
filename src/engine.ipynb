```python
import chess
import random
import chess.engine
import chess.svg
from IPython.display import display, clear_output, SVG  
import time
```

```python
def get_action(fen, value_func, mode):
    """
    Get a random action from the list of legal moves.
    """
    board = chess.Board(fen)
    if mode == "random":
        return random.choice(list(board.legal_moves))
    else:
        #print(hex(id(states_moves_values)))
        d = value_func[fen] 
        return max(d, key=d.get)
```

```python
board = chess.Board("7k/8/8/8/8/8/R7/1R5K w - - 0 1")
max_moves = 3
number_of_games = 10000


states_moves_values = {}

for _ in range(number_of_games):
    reward = -1
    state_moves_in_game = []
    
    # Save board state and moves
    board = chess.Board("7k/8/8/8/8/8/R7/1R5K w - - 0 1")
    for _ in range(max_moves):
        state = board.fen()

        if board.turn == chess.WHITE:
            move = get_action(board.fen(), "random")
            move_str = board.san(move)
            state_moves_in_game.append((state, move_str))
            
        else:
            move = random.choice(list(board.legal_moves))

        board.push(move)

        # check if game is over
        if board.is_checkmate():
            #print("Checkmate")
            reward = 10
            break
    
    for m in state_moves_in_game:
        if m[0] not in states_moves_values:
            states_moves_values[m[0]] = {}
        states_moves_values[m[0]][m[1]] = states_moves_values[m[0]].get(m[1], 0) + reward
```

## Gameplay

```python
board = chess.Board("7k/8/8/8/8/8/R7/1R5K w - - 0 1")
while not board.is_game_over():
    clear_output(wait=True)
    display(SVG(chess.svg.board(board, size=300)))  # Display the board

    if board.turn == chess.BLACK:  # Your turn (Black)
        print("Available moves:")
        legal_moves = list(board.legal_moves)
        for i, move in enumerate(legal_moves):
            print(f"{i}: {board.san(move)}")
        
        # Get your move
        move_index = int(input("Enter the number of the move you want to play: "))
        move = legal_moves[move_index]
        board.push(move)
    else:  # White's turn (program)
        move = get_action(board.fen(), states_moves_values, "best")
        board.push_san(move)

# Display the final board and result
clear_output(wait=True)
display(SVG(chess.svg.board(board, size=300)))
print("Game over!")
print("Result:", board.result())
```
